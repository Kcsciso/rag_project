```python
import os
from PIL import Image

img_dir = "data/jaka_markdown/JAKA_Manual/auto/images"
valid_imgs = 0
noise_imgs = 0

for file in os.listdir(img_dir):
    if not file.lower().endswith(('.png', '.jpg', '.jpeg')):
        continue
        
    try:
        with Image.open(os.path.join(img_dir, file)) as img:
            w, h = img.size
            # 过滤策略: 长宽小于 40 像素，或长宽比超过 15:1 (极度细长的线框)
            if w < 40 or h < 40 or (w / max(h, 1) > 15) or (h / max(w, 1) > 15):
                noise_imgs += 1
            else:
                valid_imgs += 1
    except Exception:
        pass

print(f"📊 扫描完毕！")
print(f"✅ 有效图片: {valid_imgs} 张")
print(f"🗑️ 拦截废图: {noise_imgs} 张")
```

```python
import re

def parse_table(html):
    rows = re.findall(r'<tr>(.*?)</tr>', html, re.I)
    for i, row in enumerate(rows):
        # 提取单元格并清除内部嵌套的 HTML 标签
        cells = [re.sub(r'<[^>]+>', '', c).strip() for c in re.findall(r'<t[dh]>(.*?)</t[dh]>', row, re.I)]
        print("| " + " | ".join(cells) + " |")
        
        # 打印表头底部的分隔线
        if i == 0:
            print("|" + "|".join(["---"] * len(cells)) + "|")

# 读取 Markdown 文件
path = "data/jaka_markdown/JAKA_Manual/auto/JAKA_Manual.md"
with open(path, "r", encoding="utf-8") as f:
    # 查找所有的 table 标签块
    tables = re.findall(r'<table.*?>.*?</table>', f.read(), re.I)
    
    print(f"找到 {len(tables)} 个 HTML 表格\n")
    # 仅测试前 2 个表格
    for t in tables[:2]:
        parse_table(t)
        print("-" * 40)
```

```python
import os
import re
from collections import Counter
from PIL import Image

md_path = "/home/kasm-user/rag_project/data/jaka_markdown/JAKA_Manual/auto/JAKA_Manual.md"
base_dir = os.path.dirname(md_path)

if not os.path.exists(md_path):
    print(f"❌ 找不到文件: {md_path}")
else:
    with open(md_path, "r", encoding="utf-8") as f:
        content = f.read()

    lines = content.splitlines()
    print("=" * 60)
    print("📋 【1. 基础文档统计】")
    print(f"• 文件绝对路径: {md_path}")
    print(f"• 文件大小: {len(content.encode('utf-8')) / 1024:.2f} KB")
    print(f"• 总字符数: {len(content):,} 字符")
    print(f"• 总行数: {len(lines):,} 行")

    # 1. 标题层级统计
    headers = [line.strip() for line in lines if line.strip().startswith('#')]
    header_counts = Counter(re.match(r'^(#+)', h).group(1) for h in headers if re.match(r'^(#+)', h))
    print("\n📊 【2. 标题层级分布 (Headers)】")
    for h_tag in sorted(header_counts.keys()):
        print(f"• {h_tag} ({len(h_tag)}级标题): {header_counts[h_tag]} 处")
    
    print("\n🔍 抽样前 5 个各级标题示例:")
    for h in headers[:5]:
        print(f"  - {h}")

    # 2. 表格形式统计 (HTML 表格 vs Markdown 表格)
    html_tables = re.findall(r'<table[^>]*>.*?</table>', content, re.I | re.DOTALL)
    md_table_lines = [l for l in lines if re.match(r'^\s*\|.*\|\s*$', l)]
    print("\n📊 【3. 表格格式分布】")
    print(f"• HTML <table> 标签块: {len(html_tables)} 个")
    print(f"• 原生 Markdown 表格行: {len(md_table_lines)} 行")
    if html_tables:
        print("  ⚠️ 发现 HTML 表格，需要通过 clean_html_tables 进行纯 Markdown 规整转换。")

    # 3. 多模态图片引用与物理文件存在性检测
    img_matches = list(re.finditer(r'(!\[(.*?)\]\((images/[^)]+)\))', content))
    print("\n🖼️ 【4. 多模态图片探查】")
    print(f"• Markdown 中检测到的图片引用数: {len(img_matches)} 处")
    
    existing_imgs = 0
    missing_imgs = 0
    small_or_icon_imgs = 0
    normal_imgs = 0
    img_size_samples = []

    for m in img_matches:
        rel_path = m.group(3)
        abs_img_path = os.path.join(base_dir, rel_path)
        if os.path.exists(abs_img_path):
            existing_imgs += 1
            try:
                with Image.open(abs_img_path) as img:
                    w, h = img.size
                    if w < 60 or h < 60 or (w / max(h, 1) > 10) or (h / max(w, 1) > 10):
                        small_or_icon_imgs += 1
                    else:
                        normal_imgs += 1
                        if len(img_size_samples) < 3:
                            img_size_samples.append((rel_path, f"{w}x{h}"))
            except Exception:
                pass
        else:
            missing_imgs += 1

    print(f"• 物理文件实际存在: {existing_imgs} 张")
    print(f"• 物理文件缺失: {missing_imgs} 张")
    print(f"• 图标/细长装饰线 (将被物理过滤剔除): {small_or_icon_imgs} 张")
    print(f"• 待交由 VLM 解析的高价值大图/截图: {normal_imgs} 张")
    if img_size_samples:
        print(f"  🔍 大图样本: {img_size_samples}")

    # 4. 目录检测 (TOC)
    toc_candidates = [l for l in lines if re.search(r'(?:(?:\.\s*){2,}|…)\s*\d+\s*$', l)]
    print("\n📑 【5. 目录与骨架检测】")
    print(f"• 疑似目录引导点行数: {len(toc_candidates)} 行")

    # 5. 前 30 行预览
    print("\n📄 【6. 文档前 30 行内容预览】")
    print("-" * 60)
    for l in lines[:30]:
        print(l)
    print("=" * 60)
```

```python
import json
import os
import re

CHUNKS_PATH = "/home/kasm-user/rag_project/data/jaka_manual_chunks.json"

def load_chunks():
    if not os.path.exists(CHUNKS_PATH):
        print(f"❌ 未找到切片文件: {CHUNKS_PATH}，请先运行 markdown_loader.py")
        exit(1)
    with open(CHUNKS_PATH, "r", encoding="utf-8") as f:
        return json.load(f)

def print_overview_table(chunks, limit=None):
    """打印切片总览索引表：序号 | 章节路径 | 字符数 | 图片提纯状态 | 内容摘要"""
    print("\n" + "=" * 105)
    print(f"{'序号':<6} | {'章节层级路径':<35} | {'字符数':<6} | {'图表提纯状态':<16} | {'内容摘要'}")
    print("-" * 105)
    
    total = len(chunks) if limit is None else min(limit, len(chunks))
    for idx in range(total):
        c = chunks[idx]
        path = (c['path'][:32] + "..") if len(c['path']) > 32 else c['path']
        length = len(c['content'])
        
        # 判断 VLM 提纯状态
        if "[图表参数智能提纯]" in c['content']:
            if "仅为UI示意图" in c['content'] or "仅为示意图" in c['content']:
                vlm_status = "🖼️ 纯UI示意"
            else:
                vlm_status = "✨ 提取到参数"
        else:
            vlm_status = "📝 纯文本"
            
        # 提取第一行有效文本作为摘要
        clean_text = re.sub(r'[#>\*\!\[\]\(\)]', '', c['content']).strip()
        summary = clean_text.replace('\n', ' ')[:30] + "..." if len(clean_text) > 30 else clean_text
        
        print(f"{idx + 1:<6} | {path:<35} | {length:<6} | {vlm_status:<16} | {summary}")
    print("=" * 105)

def show_chunk_detail(chunks, idx_1_based):
    """按 1-based 序号打印完整切片卡片"""
    idx = idx_1_based - 1
    if not (0 <= idx < len(chunks)):
        print(f"❌ 序号超出范围（有效范围: 1 ~ {len(chunks)}）")
        return
        
    c = chunks[idx]
    print("\n" + "=" * 80)
    print(f"📦 【切片详情卡片】 序号: #{idx_1_based} / {len(chunks)}")
    print(f"📍 章节路径: {c['path']}")
    print(f"📏 字符总数: {len(c['content'])}")
    print("-" * 80)
    print(c['content'])
    print("=" * 80 + "\n")

def search_chunks(chunks, keyword):
    """关键词搜索并显示命中章节和上下文"""
    print(f"\n🔍 正在检索关键词: '{keyword}' ...")
    matches = []
    for idx, c in enumerate(chunks):
        if keyword.lower() in c['content'].lower() or keyword.lower() in c['path'].lower():
            matches.append(idx + 1)
            
    print(f"💡 共命中 {len(matches)} 个切片: {matches}\n")
    for seq in matches:
        c = chunks[seq - 1]
        print(f"• [切片 #{seq}] 章节: {c['path']}")
        # 截取匹配词附近 60 字符
        pos = c['content'].lower().find(keyword.lower())
        start = max(0, pos - 20)
        end = min(len(c['content']), pos + 60)
        snippet = c['content'][start:end].replace('\n', ' ')
        print(f"  👉 命中片段: ...{snippet}...\n")

if __name__ == "__main__":
    chunks = load_chunks()
    
    # 1. 打印前 30 个切片目录表
    print_overview_table(chunks, limit=30)
    
    # 2. 自动检索典型工业参数关键词
    search_chunks(chunks, "192.168")
    
    # 3. 抽样查看具体切片（例如第 171 号）
    show_chunk_detail(chunks, 171)
```
